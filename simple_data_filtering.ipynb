# Simple Data Filtering

**Description:** Create a collection of student records and filter students based on marks, age, or department.

**Objective:** Understand how conditions can be used to filter structured data.

**Tools:** Python, Jupyter Notebook

## 1. Student records

Each student is a dictionary, and the full collection is a list of dictionaries.

```python
students = [
    {'name': 'Asha', 'age': 20, 'department': 'CSE', 'marks': 88},
    {'name': 'Rahul', 'age': 22, 'department': 'ECE', 'marks': 72},
    {'name': 'Meera', 'age': 19, 'department': 'CSE', 'marks': 65},
    {'name': 'Vikram', 'age': 21, 'department': 'ME', 'marks': 91},
    {'name': 'Priya', 'age': 23, 'department': 'CSE', 'marks': 79},
    {'name': 'Arjun', 'age': 20, 'department': 'ECE', 'marks': 55},
    {'name': 'Sneha', 'age': 22, 'department': 'ME', 'marks': 84},
]

print(len(students), 'students loaded')
```

## 2. Filter function

`filter_students` takes the records and a condition (a function that returns True or False for one student), and returns the students that match.

```python
def filter_students(records, condition):
    return [s for s in records if condition(s)]
```

## 3. Display helper

```python
def show(title, rows):
    print('\n{} ({} found)'.format(title, len(rows)))
    for s in rows:
        print('  {:<8} age={:<3} dept={:<4} marks={}'.format(s['name'], s['age'], s['department'], s['marks']))
```

## 4. Filtering on conditions

The deliverables ask for filtering on at least two conditions. The examples below use one condition, then combine conditions with `and` and `or`.

```python
# Condition 1: marks
high_marks = filter_students(students, lambda s: s['marks'] >= 80)
show('Marks >= 80', high_marks)

# Condition 2: department
cse_students = filter_students(students, lambda s: s['department'] == 'CSE')
show('Department = CSE', cse_students)

# Two conditions combined with AND
combined = filter_students(students, lambda s: s['marks'] >= 75 and s['age'] <= 21)
show('Marks >= 75 AND age <= 21', combined)

# Two conditions combined with OR
either = filter_students(students, lambda s: s['department'] == 'ME' or s['marks'] < 60)
show('Department = ME OR marks < 60', either)
```

## 5. Try it yourself

Change the numbers or department names above and run the cell again to see different results. Add your own filter, for example students aged 20 or below in the ECE department.

```python
# Your own filter
my_result = filter_students(students, lambda s: s['age'] <= 20 and s['department'] == 'ECE')
show('Age <= 20 AND department = ECE', my_result)
```

## 6. Interview questions

**What is data filtering?**
Selecting only the rows that match a condition, while leaving the rest of the data unchanged.

**How would you filter a Pandas DataFrame?**
Use a boolean condition inside square brackets, for example `df[df['marks'] >= 80]`. Combine conditions with `&` (and) or `|` (or), wrapping each condition in parentheses.

**What is the difference between filtering and sorting?**
Filtering removes rows that do not match a condition. Sorting keeps every row but changes their order.
